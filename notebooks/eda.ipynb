```python
# ANALISIS DE ARCHIVO CON COLUMNAS RENOMBRADAS
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import yaml
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / "config" / "config.yaml").is_file():
    ROOT = ROOT.parent
config_path = ROOT / "config" / "config.yaml"
if not config_path.is_file():
    raise FileNotFoundError("No se encontro config/config.yaml desde la carpeta actual.")
with config_path.open("r", encoding="utf-8") as file:
    config = yaml.safe_load(file)
bronze_dir = ROOT / config["paths"]["bronze_dir"]

caracteristicas_generales = pd.read_csv(
    bronze_dir / config["source"]["csv_salida_cg"], low_memory=False
)
print("Caracteristicas generales")
display(caracteristicas_generales.head(20))
display(caracteristicas_generales.shape)
display(caracteristicas_generales.info())
display(caracteristicas_generales.describe())
```

```python
# ANALISIS EXPLORATORIO DE DATOS SIN RENOMBRAR COLUMNAS
caracteristicas_generales = pd.read_csv(
    bronze_dir / config["source"]["csv_salida_cg_sinrenombrar"], low_memory=False
)
print("Caracteristicas generales sin renombrar")
display(caracteristicas_generales.head(20))
display(caracteristicas_generales.shape)
display(caracteristicas_generales.info())
display(caracteristicas_generales.describe())
```

```python
# ANALISIS EXPLORATORIO DE OCUPADOS SIN RENOMBRAR COLUMNAS
ocupados = pd.read_csv(
    bronze_dir / config["source"]["csv_salida_ocupados_sinrenombrar"], low_memory=False
)
display(ocupados.head(20))
display(ocupados.shape)
display(ocupados.info())
display(ocupados.describe())
```

```python
# ANALISIS EXPLORATORIO DE OCUPADOS CON COLUMNAS RENOMBRADAS
ocupados = pd.read_csv(
    bronze_dir / config["source"]["csv_salida_ocupados"], low_memory=False
)
display(ocupados.head(20))
display(ocupados.shape)
display(ocupados.info())
display(ocupados.describe())
```

```python
# detectar nulos
print("Cantidad de nulos en caracteristicas generales")
display(caracteristicas_generales.isnull().sum())
print("Cantidad de nulos en ocupados")
display(ocupados.isnull().sum())
```
