```python
#ANALISIS DE ARCHIVO C.GENERALES CON COLUMNAS RENOMBRADAS
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import yaml
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / "config" / "config.yaml").is_file():
    ROOT = ROOT.parent
config_path = ROOT / "config" / "config.yaml"
if not config_path.is_file():
    raise FileNotFoundError("No se encontro config/config.yaml desde la carpeta actual.")
with config_path.open("r", encoding="utf-8") as file:
    config = yaml.safe_load(file)
bronze_dir = ROOT / config["paths"]["bronze_dir"]
silver_dir = ROOT / config["paths"]["silver_dir"]

caracteristicas_generales_ocupados = pd.read_excel(silver_dir / config["source"]["excel_salida_datasets_traducido_clean"])
print("Caracteristicas generales")
display(caracteristicas_generales_ocupados.head(20))
display(caracteristicas_generales_ocupados.shape)
display(caracteristicas_generales_ocupados.info())    
display(caracteristicas_generales_ocupados.describe())


```
